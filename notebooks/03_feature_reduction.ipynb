# ClimateGuard AI - Feature Reduction

This notebook removes duplicate unit-conversion columns from the cleaned weather dataset before feature engineering.

Input: `../data/processed/02_cleaned_data.csv`
Output: `../data/processed/03_feature_reduced_data.csv`

## 1. Import Libraries

```python
from pathlib import Path

import pandas as pd
```

## 2. Load Cleaned Data

```python
input_path = Path('../data/processed/02_cleaned_data.csv')
df = pd.read_csv(input_path)

print(f'Loaded data from {input_path}')
print(f'Initial shape: {df.shape}')
display(df.head())
```

## 3. Remove Duplicate Unit Columns

```python
columns_to_remove = [
    'temperature_fahrenheit',
    'wind_mph',
    'pressure_in',
    'precip_in',
    'feels_like_fahrenheit',
    'visibility_miles',
    'gust_mph',
]

missing_columns = [column for column in columns_to_remove if column not in df.columns]
if missing_columns:
    raise ValueError(f'Required columns not found: {missing_columns}')

reduced_df = df.drop(columns=columns_to_remove)

assert not set(columns_to_remove).intersection(reduced_df.columns)
print(f'Removed columns: {columns_to_remove}')
print(f'Reduced shape: {reduced_df.shape}')
display(reduced_df.head())

# Save the reduced dataset
output_path = '../data/processed/03_feature_reduced_data.csv'
reduced_df.to_csv(output_path, index=False)

print(f"\nReduced dataset saved to: {output_path}")
```
