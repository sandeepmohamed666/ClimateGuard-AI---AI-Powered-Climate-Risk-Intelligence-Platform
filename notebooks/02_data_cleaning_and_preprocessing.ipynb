# ClimateGuard AI v3.0 - Data Cleaning and Preprocessing 

This notebook covers data cleaning, handling missing values, and preprocessing the weather dataset.

## 1. Import Libraries  

```python
import pandas as pd
import numpy as np
import sys
import os

# Add src to path
sys.path.append(os.path.abspath(os.path.join(os.path.dirname('__file__'), '..')))

from src.data_preprocessing.data_cleaner import DataCleaner

import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully")
```

## 2. Load Data from Previous Step 

```python
# Load the raw data
df = pd.read_csv('../data/processed/01_raw_data.csv')
print(f"Data loaded: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
```

## 3. Initialize Data Cleaner 

```python
cleaner = DataCleaner(df)
print("Data cleaner initialized")
```

## 4. Remove Duplicates 

```python
print("Before removing duplicates:")
print(f"Shape: {cleaner.df.shape}")
print(f"Duplicates: {cleaner.df.duplicated().sum()}")

cleaner.remove_duplicates()

print("\nAfter removing duplicates:")
print(f"Shape: {cleaner.df.shape}")
```

## 5. Handle Missing Values 

```python
# Check missing values before
print("Missing values before imputation:")
missing_before = cleaner.df.isnull().sum()
print(missing_before[missing_before > 0])

# Impute missing values using mean for numerical columns
cleaner.handle_missing_values(strategy='mean')

print("\nMissing values after imputation:")
missing_after = cleaner.df.isnull().sum()
print(missing_after[missing_after > 0])
```

## 6. Remove Impossible Weather Values 

```python
print("Removing impossible weather values...")
cleaner.remove_impossible_values()
print(f"Shape after removing impossible values: {cleaner.df.shape}")
```

## 7. Detect Outliers (IQR Method) 

```python
# Detect outliers in numerical columns
numerical_cols = cleaner.df.select_dtypes(include=[np.number]).columns

outlier_info = {}
for col in numerical_cols[:10]:  # Check first 10 numerical columns
    outliers = cleaner.detect_outliers_iqr(col)
    outlier_info[col] = outliers.sum()
    print(f"{col}: {outliers.sum()} outliers")
```

## 8. Detect Outliers (Z-Score Method) 

```python
# Detect outliers using Z-score
for col in numerical_cols[:10]:
    outliers = cleaner.detect_outliers_zscore(col, threshold=3)
    print(f"{col} (Z-score): {outliers.sum()} outliers")
```

## 9. Get Cleaned Data 

```python
df = cleaner.get_cleaned_data()
print(f"Final cleaned data shape: {df.shape}")
print(f"\nData types:")
print(df.dtypes)
```

## 10. Summary Statistics After Cleaning 

```python
print("Summary statistics after cleaning:")
display(df.describe())
```

## 11. Save Cleaned Data 

```python
# Save cleaned data
df.to_csv('../data/processed/02_cleaned_data.csv', index=False)
print("\nCleaned data saved to ../data/processed/02_cleaned_data.csv")
```

## Summary 

In this notebook, we:

2. Removed duplicate rows
3. Handled missing values using mean imputation
4. Removed impossible weather values
5. Detected outliers using IQR and Z-score methods
6. Saved the cleaned data for feature engineering
