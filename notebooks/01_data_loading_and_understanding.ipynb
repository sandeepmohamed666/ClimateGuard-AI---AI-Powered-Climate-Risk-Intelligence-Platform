# ClimateGuard AI v3.0 - Data Loading and Understanding

This notebook covers loading the historical weather dataset and performing initial data understanding.

## 1. Import Libraries 

```python
import pandas as pd
import numpy as np
import sys
import os

# Add src to path
sys.path.append(os.path.abspath(os.path.join(os.path.dirname('__file__'), '..')))

from src.data_collection.historical_loader import HistoricalDataLoader
from src.data_preprocessing.data_understanding import DataUnderstanding

import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully")
```

## 2. Load Historical Weather Data 

```python
# Define data path
data_path = '../data/raw/IndianWeatherRepository.csv'

# Load data
loader = HistoricalDataLoader(data_path)
df = loader.load_data()

print(f"Data loaded successfully!")
print(f"Shape: {df.shape}")
```

## 3. Basic Data Information 

```python
# Get data info
info = loader.get_data_info()
print("\n" + "="*50)
print("DATA INFORMATION")
print("="*50)
print(f"Rows: {info['shape'][0]}")
print(f"Columns: {info['shape'][1]}")
mem = info.get('memory_usage', info.get('memory_usage_mb'))
if mem is None:
    print("Memory Usage: N/A")
else:
    print(f"Memory Usage: {mem:.2f} MB")
print(f"Missing Values: {info['missing_values']}")
print(f"Duplicate Rows: {info['duplicate_rows']}")
```

## 4. Display First Few Rows 

```python
print("\nFirst 5 rows:")
display(df.head())
```

## 5. Data Types

```python
print("\nData Types:")
print(df.dtypes)
```

## 6. Statistical Summary

```python
print("\nStatistical Summary:")
display(df.describe())
```

## 7. Missing Values Analysis 

```python
print("\nMissing Values per Column:")
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Count': missing, 'Percentage': missing_pct})
display(missing_df[missing_df['Count'] > 0].sort_values('Count', ascending=False))
```

## 8. Unique Regions

```python
# Find region columns
region_cols = [col for col in df.columns if 'region' in col.lower() or 'city' in col.lower() or 'state' in col.lower()]
print(f"Region-related columns: {region_cols}")

if region_cols:
    for col in region_cols:
        if col in df.columns:
            print(f"\n{col} - Unique values: {df[col].nunique()}")
            print(f"Top 10 values:")
            print(df[col].value_counts().head(10))
```

## 9. Comprehensive Data Understanding Report

```python
# Generate comprehensive report
analyzer = DataUnderstanding(df)
report = analyzer.generate_comprehensive_report()

print("\n" + "="*50)
print("COMPREHENSIVE DATA UNDERSTANDING REPORT")
print("="*50)

print(f"\nDataset Shape: {report.get('dataset_shape', report.get('shape', 'N/A'))}")
# Derive column types from the dataframe since the report dict
# doesn't contain 'numerical_columns' / 'categorical_columns' keys
report['numerical_columns'] = list(df.select_dtypes(include=[np.number]).columns)
report['categorical_columns'] = list(df.select_dtypes(exclude=[np.number]).columns)

print(f"\nNumerical Columns: {len(report['numerical_columns'])}")
print(f"Categorical Columns: {len(report['categorical_columns'])}")
total_missing = sum(report.get('missing_values', {}).values())
print(f"\nTotal Missing Values: {total_missing}")
print(f"Duplicate Rows: {report['duplicate_rows']}")
```

## 10. Save Cleaned Data for Next Steps

```python
# Save the raw data for next steps
df.to_csv('../data/processed/01_raw_data.csv', index=False)
print("\nData saved to ../data/processed/01_raw_data.csv")
```
