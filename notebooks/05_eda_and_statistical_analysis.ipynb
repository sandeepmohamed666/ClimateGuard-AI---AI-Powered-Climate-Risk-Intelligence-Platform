# ClimateGuard AI v3.0 - EDA and Statistical Analysis

This notebook covers  exploratory data analysis and statistical analysis of the weather dataset.

## 1. Import Libraries 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# Add src to path
sys.path.append(os.path.abspath(os.path.join(os.path.dirname('__file__'), '..')))

from src.eda.exploratory_analysis import ExploratoryDataAnalysis
from src.eda.statistical_analysis import StatisticalAnalysis

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 8)

import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully")
```

## 2. Load Engineered Data 

```python
# Load engineered data
df = pd.read_csv('../data/processed/04_engineered_data.csv')
print(f"Data loaded: {df.shape}")
print(f"Columns: {len(df.columns)}")
```

## 3. Initialize EDA and Statistical Analysis 

```python
eda = ExploratoryDataAnalysis(df)
stats = StatisticalAnalysis(df)
print("EDA and Statistical Analysis initialized")
```

## 4. Univariate Distributions 

```python
# Plot distributions for key numerical columns
key_cols = eda.numerical_cols[:6]
print(f"Plotting distributions for: {key_cols}")
os.makedirs('reports', exist_ok=True)
eda.plot_univariate_distributions(columns=key_cols)
print("Univariate distributions saved to reports/")
```

## 5. Boxplots for Outlier Detection 

```python
eda.plot_boxplots(columns=key_cols)
print("Boxplots saved to reports/")
```

## 6. Correlation Heatmap 

```python
eda.plot_correlation_heatmap()
print("Correlation heatmap saved to reports/")
```

## 7. Scatter Matrix 

```python
eda.plot_scatter_matrix(columns=key_cols[:4])
print("Scatter matrix saved to reports/")
```

## 8. Time Series Analysis 

```python
# Check for date column
date_cols = [col for col in df.columns if 'date' in col.lower()]
if date_cols:
    eda.plot_time_series(date_cols[0], key_cols[:3])
    print("Time series plot saved to reports/")
else:
    print("No date column found for time series analysis")
```

## 9. Monthly Aggregates 

```python
if date_cols:
    eda.plot_monthly_aggregates(date_cols[0], key_cols[:3])
    print("Monthly aggregates plot saved to reports/")
```

## 10. Regional Analysis 

```python
# Check for region column
region_cols = [col for col in df.columns if 'region' in col.lower() or 'city' in col.lower()]
if region_cols and key_cols:
    eda.plot_regional_analysis(region_cols[0], key_cols[0])
    print("Regional analysis plot saved to reports/")
```

## 11. Statistical Analysis - Descriptive Statistics 

```python
desc_stats = stats.calculate_descriptive_stats()
print("Descriptive Statistics:")
display(desc_stats)
```

## 12. Correlation Matrix 

```python
corr_matrix = stats.calculate_correlation()
print("Correlation Matrix:")
display(corr_matrix)
```

## 13. Confidence Intervals 

```python
print("Confidence Intervals (95%):")
for col in stats.numerical_cols[:5]:
    ci = stats.calculate_confidence_interval(col)
    print(f"{col}: [{ci[0]:.4f}, {ci[1]:.4f}]")
```

## 14. Normality Tests  

```python
print("Normality Tests (Shapiro-Wilk):")
for col in stats.numerical_cols[:5]:
    test = stats.perform_normality_test(col)
    print(f"{col}: p-value = {test['p_value']:.4f}, Normal = {test['is_normal']}")
```

## 15. Outlier Detection Summary 

```python
print("Outlier Detection Summary:")
for col in stats.numerical_cols[:5]:
    outliers_zscore = stats.detect_outliers_statistical(col, 'zscore')
    outliers_iqr = stats.detect_outliers_statistical(col, 'iqr')
    print(f"{col}: Z-score outliers = {outliers_zscore.sum()}, IQR outliers = {outliers_iqr.sum()}")
```

## 16. Generate Comprehensive Statistical Report

```python
stat_report = stats.generate_statistical_report()
stats.print_report()
```
